````python
# 1. Lista de aprendices
aprendices = ["Ana", "Carlos", "María", "Luis", "Sofía"]
print("Lista de aprendices:", aprendices)
print("Cantidad de aprendices:", len(aprendices))

# 2. Diccionario para representar un aprendiz
aprendiz = {
    "documento": "1001",
    "nombre": "Ana",
    "programa": "ADSO",
    "nota_python": 4.2,
    "asistencia": 92
}
print("\nDiccionario aprendiz:", aprendiz)
print("Nombre:", aprendiz["nombre"])
print("Nota Python:", aprendiz["nota_python"])

# 3. Lista de diccionarios (registros iniciales)
datos_aprendices = [
    {"documento": "1001", "nombre": "Ana", "programa": "ADSO", "nota_python": 4.2, "asistencia": 92},
    {"documento": "1002", "nombre": "Carlos", "programa": "ADSO", "nota_python": 3.8, "asistencia": 85},
    {"documento": "1003", "nombre": "María", "programa": "ADSO", "nota_python": 4.7, "asistencia": 96},
]

# RETO DEL APRENDIZ (Sección 3.2):
# Agregar dos registros nuevos a datos_aprendices con todos sus campos
datos_aprendices.append({
    "documento": "1004", "nombre": "Luis", "programa": "ADSO", "nota_python": 2.9, "asistencia": 70
})
datos_aprendices.append({
    "documento": "1005", "nombre": "Sofía", "programa": "ADSO", "nota_python": 5.0, "asistencia": 98
})

print("\nLista de diccionarios actualizada (con reto completado):")
datos_aprendices
````

````python
import numpy as np

# 1. Crear arreglo NumPy con las notas
notas = np.array([4.2, 3.8, 4.7, 2.9, 5.0])

print("--- Estadísticas Básicas ---")
print("Notas:", notas)
print("Promedio:", np.mean(notas))
print("Mediana:", np.median(notas))
print("Nota Máxima:", np.max(notas))
print("Nota Mínima:", np.min(notas))

# 2. Operación vectorizada: Bonificación controlada (+0.2 sin superar 5.0)
notas_bonificadas = np.clip(notas + 0.2, 0, 5)
print("\n--- Operación Vectorizada ---")
print("Notas originales:", notas)
print("Notas bonificadas:", notas_bonificadas)

# 3. Filtro de aprobación (notas >= 3.5)
notas_aprobadas = notas[notas >= 3.5]
porcentaje_aprobacion = len(notas_aprobadas) / len(notas) * 100

print("\n--- Filtro y Porcentaje ---")
print("Notas aprobadas:", notas_aprobadas)
print(f"Porcentaje de aprobación: {porcentaje_aprobacion:.1f}%")
````

````python
import pandas as pd

# 1. Crear una Serie en Pandas a partir de una lista
serie_notas = pd.Series(
    data=[4.2, 3.8, 4.7, 2.9, 5.0],
    index=["Ana", "Carlos", "María", "Luis", "Sofía"]
)
print("--- Serie de Notas ---")
print(serie_notas)

# 2. Crear un DataFrame desde la lista de diccionarios (datos_aprendices del Paso 2)
df_aprendices = pd.DataFrame(datos_aprendices)

# 3. Agregar columna calculada: estado (Aprobado / En riesgo)
df_aprendices["estado"] = np.where(
    df_aprendices["nota_python"] >= 3.5,
    "Aprobado",
    "En riesgo"
)

# RETO DEL APRENDIZ (Sección 3.4):
# Crear columna alerta_asistencia: si asistencia < 80 -> "Revisar asistencia", sino -> "Sin alerta"
df_aprendices["alerta_asistencia"] = np.where(
    df_aprendices["asistencia"] < 80,
    "Revisar asistencia",
    "Sin alerta"
)

print("\n--- DataFrame Final con Columnas Calculadas ---")
````
